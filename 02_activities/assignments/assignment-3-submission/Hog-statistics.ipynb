Import packages

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
# This needs to be run to open the excel file
```

```python
hogs = pd.read_csv(r"C:\Users\tdudg\Documents\DSI\visualization\02_activities\assignments\hogs.csv")
hogs.head()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    hogs["Dressed Weight (kg)"],
    hogs["Estimated Market Hog Value ($)"],
    alpha=0.6
)

plt.xlabel("Dressed Weight (kg)")
plt.ylabel("Estimated Market Hog Value ($)")
plt.title("Hog value vs Weight")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    hogs["Total Cost per Pig ($)"],
    hogs["Net Return ($)"],
    alpha=0.6
)

plt.xlabel("Total Cost per Pig ($)")
plt.ylabel("Net Return ($)")
plt.title("Hog Cost vs Return")

plt.tight_layout()
plt.show()
```
